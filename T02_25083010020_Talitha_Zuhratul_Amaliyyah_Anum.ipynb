```python
import numpy as np
import matplotlib.pyplot as plt

def simpson_13(y, dx):
    """
    Menghitung integrasi numerik menggunakan Aturan Simpson 1/3.
    y  : array nilai (luas penampang)
    dx : lebar interval grid (dalam meter)
    """
    n = len(y) - 1
    if n % 2 != 0:
        raise ValueError("Jumlah interval (n) harus genap! Jumlah titik data harus ganjil.")

    integral = y[0] + y[-1]
    for i in range(1, n):
        if i % 2 == 1:
            integral += 4 * y[i]
        else:
            integral += 2 * y[i]

    return (dx / 3.0) * integral

print("✓ Cell 1 Berhasil: Fungsi Simpson 1/3 siap digunakan.")
```

```python
# --- PARAMETER SKALA GRID ---
# 1 dash-kosong = 5 cm + 5 cm = 10 cm = 0.1 meter
dx = 0.1  # meter

# --- DATA HASIL PEMBACAAN GRID (17 TITIK STASIUN) ---
# Lebar lambung diukur dari Tampak Atas (dalam meter)
lebar = np.array([
    0.00, 0.12, 0.22, 0.30, 0.36, 0.40, 0.42, 0.43,
    0.43, 0.42, 0.40, 0.36, 0.30, 0.22, 0.14, 0.06, 0.00
])

# Tinggi/Kedalaman lambung diukur dari Tampak Samping (dalam meter)
tinggi = np.array([
    0.00, 0.18, 0.26, 0.32, 0.36, 0.38, 0.40, 0.40,
    0.40, 0.40, 0.38, 0.36, 0.32, 0.26, 0.18, 0.10, 0.00
])

print(f"Jumlah titik stasiun (N) : {len(lebar)} titik")
print(f"Jumlah interval segmen (n): {len(lebar) - 1} segmen (genap)")
print(f"Lebar interval (dx)       : {dx} meter")
```

```python
# Karena lantai kapal datar (tanpa keel), penampang melintang berbentuk persegi panjang: A = w * h
luas_penampang = lebar * tinggi

print("=" * 45)
print(" STASIUN | LEBAR (m) | TINGGI (m) | LUAS A(x) (m²)")
print("=" * 45)
for i in range(len(lebar)):
    print(f"  x_{i:02d}   |   {lebar[i]:.2f}    |    {tinggi[i]:.2f}    |   {luas_penampang[i]:.4f}")
print("=" * 45)
```

```python
# Hitung volume 1 lambung menggunakan integrasi Simpson 1/3
vol_satu_lambung_m3 = simpson_13(luas_penampang, dx)

# Katamaran memiliki 2 lambung simetris
vol_total_m3 = 2 * vol_satu_lambung_m3

# Konversi ke Liter (1 m³ = 1000 Liter)
vol_satu_lambung_liter = vol_satu_lambung_m3 * 1000
vol_total_liter = vol_total_m3 * 1000

print("=" * 55)
print("         HASIL PERHITUNGAN VOLUME KAPAL KATAMARAN")
print("=" * 55)
print(f"Volume 1 Lambung         : {vol_satu_lambung_m3:.6f} m³  ({vol_satu_lambung_liter:.2f} Liter)")
print(f"Volume Total (2 Lambung) : {vol_total_m3:.6f} m³  ({vol_total_liter:.2f} Liter)")
print("=" * 55)
```
